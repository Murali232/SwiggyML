```python
results=[]
cont_cols=[col for col in df.columns if df[col].dtype=='int64'or df[col].dtype=='float64']
cont_cols.remove('is_peak_hour','is_gated_community','is_pure_veg_restaurant','is_swiggy_one_member','is_public_holiday','delivery_lat')
for col in cont_cols:
    hvo=[]#high value outliers
    lvo=[]#low value outliers
    q1=df[col].describe()['25%']
    q3=df[col].describe()['75%']

    iqr=q3-q1

    for val in df[col].values:
        if val>q3+(1.5*iqr):
            hvo.append(val)
        if val<q1-(1.5*iqr):
            lvo.append(val)
    
    results.append({
        'column' : col,
        'high_Outliers': len(hvo),
        'low_Outliers':len(lvo)
        })
pd.DataFrame(results)
```

```python
cont_cols=[col for col in df.columns if df[col].dtype=='int64'or df[col].dtype=='float_64']
cat_cols=[col for col in df.columns if df[col].dtype=='str'or df[col].dtype=='object']
```

```python
cont_cols
```

```python
cat_cols
```

```python
if 'delivered_timestamp' in cat_cols:
    cat_cols.remove('delivered_timestamp')
```

```python
for col in cont_cols:
    x_train[col] = x_train[col].fillna(x_train[col].mean())
    x_test[col] = x_test[col].fillna(x_train[col].mean())
    
for col in cat_cols:
    
    x_train[col]=x_train[col].fillna(x_train[col].mode()[0])
    x_test[col]=x_test[col].fillna(x_train[col].mode()[0])
```

```python
from sklearn.preprocessing import StandardScaler,OneHotEncoder
import numpy as np

for col in cont_cols:
    scaler=StandardScaler()
    x_train[col]=scaler.fit_transform(np.array(x_train[col]).reshape(-1,1))
    x_test[col]=scaler.transform(np.array(x_test[col]).reshape(-1,1))
oe = OneHotEncoder(handle_unknown='infrequent_if_exist')
x_train_encoded= oe.fit_transform(x_train[cat_cols])
x_test_encoded=oe.transform(x_test[cat_cols])
```

```python
x_train_encoded_df=pd.DataFrame(x_train_encoded.toarray(),columns=oe.get_feature_names_out(cat_cols),index=x_train.index)
x_test_encoded_df=pd.DataFrame(x_test_encoded.toarray(),columns=oe.get_feature_names_out(cat_cols),index=x_test.index)
```

```python
x_train_final=pd.concat([x_train[cont_cols],x_train_encoded_df],axis=1)
x_test_final=pd.concat([x_test[cont_cols],x_test_encoded_df],axis=1)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

linreg=LinearRegression()
linreg.fit(x_train_final,y_train)
```

```python
y_pred=linreg.predict(x_test_final)
```

```python
r2_score(y_test,y_pred)
```
